```python
def calculate_area(scores,message="Scores for 1 Over"):
    total = 0
    for score in scores:
        total += score
    return total / len(scores), message
```

```python
score,message = calculate_area([1,0,5,6,3,2],message="Scores for 1 Over")

print(f"Score is {round(score, 2)} and message is {message}")
```

```python
def evaluate_model(name,score,threshold=0.7):
    if score >= threshold:
        return f"Model {name} passed with score {score}"
    else:
        return f"Model {name} failed with score {score}"
```

```python
print(evaluate_model("ollama",0.8))
print(evaluate_model("agy",0.6))
print(evaluate_model("mistrial",0.72))
```

### Objects and Classes in Python

```python
class AIModel:

    def __init__(self,name,score,local):
        self.name = name
        self.score = score
        self.local = local

    def is_good_model(self):
        if self.score >= 0.8:
            return True
        else:
            return False
```

```python
obj_1 = AIModel("ollama",0.9,True)
print(obj_1.is_good_model())
obj_2 = AIModel("agy",0.6,False)
print(obj_2.is_good_model())
obj_3 = AIModel("mistrial",0.72,True)
print(obj_3.is_good_model())
```

```python
text = """
Artificial intelligence systems use models.
Models process data and models generate outputs.
Applied AI engineers connect models to useful applications.
"""
```

```python
def analyze_text(text):

    # Text conversion to lower case
    text = text.lower()

    # Now let's remove the . from the text
    for letter in text:
        if letter == ".":
            text = text.replace(letter,"")

    # split into individual words
    text = text.split(" ")

    count = len(text)

    # count how often repeat words occur
    word_count = {}
    for word in text:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    return text,count,word_count
```

```python
print(analyze_text(text))
```

```python
class ModelEvaluator:

    def __init__(self, threshold=0.7):
        self.threshold = threshold

    def evaluate(self, scores):
        passed = []

        for score in scores:
            if score >= self.threshold:
                passed.append(score)

        return passed


try:
    evaluator = ModelEvaluator(0.8)

    results = evaluator.evaluate(
        [0.92, 0.54, 0.83]
    )

    print(results)

except Exception as error:
    print(error)
```

# Modern Python Patterns


```python
def add(a,b):
    return a+b

add(5,6)
```

```python
def add_multiple(*args):
    print(args)
    print(type(args))
```

```python
add_multiple(1,2,3,4,5)
```

```python
def average_score(*scores):
    return sum(scores) / len(scores)

result  = average_score(0.8,0.9,0.7,0.6,0.5)
print(result)
```

```python
def show_config(**kwargs):
    print(kwargs)
    print(type(kwargs))
```

```python
show_config(model="ollama",score=0.9,local=True)
```

```python
def create_model_config(**config):
    return config

settings = create_model_config(model="ollama",score=0.9,local=True)
```
